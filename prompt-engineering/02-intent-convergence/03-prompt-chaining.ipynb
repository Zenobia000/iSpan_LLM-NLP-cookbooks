# 任務分解與鏈式工作流(Prompt Chaining)

## 模組脈絡:從「一段 prompt」到「人類掌舵的工作流」

本筆記隸屬 **02-意圖收斂**。與其把所有要求塞進單一巨量 prompt,不如把任務拆成**可驗證的多個階段**,上一階段的輸出餵給下一階段。這有兩個定位:

- **spec 分解的前奏**:每個階段其實是一小片 spec(見 04);把大目標切成可檢查的步驟。
- **agent harness(M5)的橋**:chaining 是人類預先設計好的固定流程;當「下一步由模型自行決定」時,它就升級成 agent loop。

> 注意:這裡的「步驟」是**工作流層的狀態轉移**(人類設計),不是要模型在單次回覆裡「一步一步想」(那對推理模型已非必要,見 02)。

```python
import os
from openai import OpenAI
client = OpenAI()  # 讀取 OPENAI_API_KEY
OPENAI_MODEL = os.getenv("OPENAI_MODEL", "gpt-5.4-mini")

def get_completion(messages, model=OPENAI_MODEL, temperature=0, max_tokens=1000, format_type=None):
    kwargs = dict(model=model, input=messages, temperature=temperature, max_output_tokens=max_tokens)
    if format_type:
        kwargs["text"] = {"format": {"type": format_type}}
    try:
        return client.responses.create(**kwargs).output_text
    except Exception as e:
        return f"[API 錯誤] {type(e).__name__}: {e}"
```

## 人類都不可能一次到位，一次到位的品質可能會比較差，

生成式模型本質上就是一個文字接龍，盡可能不要跳躍性思考，而是階段性的思考，讓模型有階段性的思考推理過程，能得到更好的結果。

好的品質需要精雕細琢: 費時費力，但是品質會更好
(latency 跟 tokens 成本也比較高)

## 進階:用程式化 chaining 做「分而治之」

舊教材用「Let's break down this problem」這類咒語要模型自己分解(least-to-most)。2026 改用**人類掌舵的程式化 chaining**:明確定義階段(分解→逐步求解→彙整),每階段可檢查、可重試。

```python
# 步驟由人類明確設計(工作流),不是 prompt 咒語
question = (
    "一間咖啡廳每天賣 120 杯,成本每杯 35 元、售價 80 元,月租 6 萬,"
    "以 30 天計,月淨利約多少?"
)

# 階段 1:分解子問題
subqs = get_completion([{"role": "user", "content":
    f"把問題分解成 2-3 個可獨立計算的子問題,只列子問題不要計算:\n{question}"}])
print("【子問題】\n", subqs)

# 階段 2:把上一階段輸出餵入,逐步求解並彙整
answer = get_completion([{"role": "user", "content":
    f"依下列子問題逐一計算,最後彙整出答案:\n問題:{question}\n子問題:\n{subqs}"}])
print("\n【求解】\n", answer)
# 重點:這是人類設計的狀態轉移(工作流),正是 M5 agent loop 的前身。
```

## 本章小結

1. **chaining = 多階段工作流**:每階段一個子目標,上一階段輸出餵下一階段,階段可驗證、可重試。
2. **翻譯案例**:粗翻 → 潤飾,品質優於單輪(代價是 latency / token)。
3. **分而治之**用程式化 chaining(人類掌舵),取代舊的 least-to-most 咒語。
4. 每階段是一小片 **spec**(見 04);當「下一步由模型自行決定」,chaining 就升級為 **M5 的 agent loop**。
